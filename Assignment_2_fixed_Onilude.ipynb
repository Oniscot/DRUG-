```python
df_no_smiles=df4_clean.drop(columns='canonical_smiles')
```

```python
smiles =[]

for i in df4.canonical_smiles.tolist():
  cpd = str(i).split('.')
  cpd_head = cpd[0]
  cpd_longes = '.'.join(cpd[1:])
  smiles.append(cpd_head)
  smiles.append(cpd_longes)

Smiles=pd.Series(smiles, name="canonical_smiles")
df_clean_smiles = pd.concat([df_no_smiles, Smiles], axis = 1)
df= df_clean_smiles
```

```python
def calculate_lipinski_descriptors (smiles, verbose=False):
  moldata= []
  for elem in smiles:
    mol=Chem.MolFromSmiles(elem)
    # Handle cases where SMILES might be invalid and Chem.MolFromSmiles returns None # I used different method here because my initial syntax was not running#
    if mol is None:
        moldata.append(None)
    else:
        moldata.append(mol)

  baseData_rows= [] # Use a list to append rows, then convert to numpy array at the end

  for mol in moldata:
    if mol is not None: # Only process valid molecules
        desc_MolWt = Descriptors.MolWt(mol)
        desc_MolLogP = Descriptors.MolLogP(mol)
        desc_NumHDon = Chem.Lipinski.NumHDonors(mol) # Use Chem.Lipinski to avoid shadowing
        desc_NumHAccept = Chem.Lipinski.NumHAcceptors(mol)
        row = [desc_MolWt, desc_MolLogP, desc_NumHDon, desc_NumHAccept]
        baseData_rows.append(row)
    else:
        # For invalid SMILES, append NaNs
        baseData_rows.append([np.nan, np.nan, np.nan, np.nan])

  ColumnNames=["MW","LogP","NumHDon","NumHAccept"]
  if not baseData_rows: # If no valid molecules were processed
      descriptors = pd.DataFrame(columns=ColumnNames)
  else:
      descriptors = pd.DataFrame(data=baseData_rows, columns=ColumnNames)

  return descriptors
```

```python
df_lipinski = calculate_lipinski_descriptors(df.canonical_smiles)
df_lipinski.head()
df_lipinski.shape
```

```python
def mannwhitney (descriptor, df_combined):
  from numpy.random import seed
  from scipy.stats import mannwhitneyu
  import pandas as pd

  # Filter data for active and inactive compounds for the given descriptor
  selection = [descriptor, 'bioactivity_class']
  df = df_combined[selection]
  active = df[df['bioactivity_class'] == 'active'][descriptor]
  inactive = df[df['bioactivity_class'] == 'inactive'][descriptor]

  # Perform Mann-Whitney U test
  stat, p = mannwhitneyu(active, inactive)

  print(f"Descriptor: {descriptor}")
  print(f"Statistics={stat:.3f}, p={p:.3f}")

  # Interpret the result
  alpha = 0.05
  if p > alpha:
       interpretation = "Same distribution (fail to reject Ho)"
  else:
       interpretation = "Different distribution (reject Ho)"
  print(f"p={p:.3f}; {interpretation}")

  # Store the result in a DataFrame
  results = pd.DataFrame({"Descriptor":descriptor,
                          "Statistics":stat,
                          "p":p,
                          "alpha":alpha,
                          "Interpretation":interpretation},
                          index=[0])
  # Return the results DataFrame
  return results

# Example usage: Call the mannwhitney function for the 'MW' descriptor
# This will also save the result to a CSV file if desired inside the function
# However, the original code had 'results.to.csv' at global scope, so let's adjust.
# The corrected function will return the results, and saving can be done after calling.

# Let's call the function for 'MW' and 'LogP' to demonstrate
mw_test_results = mannwhitney('MW', df_combined)
logp_test_results = mannwhitney('LogP', df_combined)

# Combine results and save to a CSV file
all_test_results = pd.concat([mw_test_results, logp_test_results], ignore_index=True)
filename = 'mannwhitneyu_all_descriptors.csv'
all_test_results.to_csv(filename, index=False)
print(f"Mann-Whitney U test results saved to {filename}")
```
