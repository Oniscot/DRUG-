```python
from chembl_webresource_client.new_client import new_client
```

```python
target = new_client.target
```

```python
target_query=target.search("TP53")
```

```python
targets = pd.DataFrame.from_dict(target_query)
targets.head()
```

```python
selected_target=targets.target_chembl_id[0]
selected_target
```

```python
activity = new_client.activity
```

```python
results=activity.filter(target_chembl_id=selected_target).filter(standard_type="IC50")
```

```python
df1=pd.DataFrame.from_dict(results)
display(df1.head())
print(f'Shape of df1: {df1.shape}')
```

```python
df1["standard_type"].isna().sum()
```

```python
df2=df1[df1.standard_value.notna()]
```

```python
df2.head()
```

```python
bioactivity_class=[]
```

```python
for value in df2.standard_value:
  value=float(value)
  if value>=10000:
    bioactivity_class.append("inactive")
  elif value<=1000:
    bioactivity_class.append("active")
  else:
    bioactivity_class.append("intermediate")
```

```python
molecule_ids=df2.molecule_chembl_id.tolist()
canonical_smiles=df2.canonical_smiles.tolist()
standard_value=df2.standard_value.tolist()

```

```python
data=list(zip(molecule_ids,canonical_smiles,bioactivity_class,standard_value))
```

```python
df3=pd.DataFrame(data,columns=["molecule_chembl_id","canonical_smiles","bioactivity_class","standard_value"])
```

```python
df3.head()
```

```python
df3 = df3.dropna(subset=['canonical_smiles'])
df3 = df3[df3['canonical_smiles'].str.lower() != "none"]
df3 = df3[df3['canonical_smiles'].str.strip() != ""]
df3.head()
```

```python
df3.to_csv("bioactivity_preprocessed_data.csv", index=False)
```
